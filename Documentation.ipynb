# HealthConnect Final Analytics & Decision Support Package

## Executive Summary

HealthConnect Clinic loses nearly half of its scheduled capacity to no-shows (48.46% across 5,000 visits). Four weeks of analysis identified three statistically significant behavioral drivers — booking lead time, prior no-show history, and distance to clinic — and uncovered a critical insight during Week 7 testing: reminder messages only reduce no-shows for one specific risk segment (Tier 3, moderate-to-high risk), not the population as a whole. Acting on these findings, a combined policy of capping long-lead bookings and selectively overbooking high-risk slots could reclaim 1,048 patient visits, cutting the no-show rate to 27.50% while redirecting reminder spend away from segments where it has no effect.

## Final KPIs

| KPI | Finding |
|---|---|
| Booking Lead Time | No-show rate climbs from 24.84% (0–3 days out) to 60.49% (31–60 days out) — the strongest single driver |
| Prior No-Show History | Rises from 43.51% (no prior misses) to 68.82% (3+ prior misses) |
| Distance to Clinic | Flat (46–49%) under 20km, jumps to 57.76% beyond 20km |
| Reminder Channel Efficacy | SMS outperforms Email and WhatsApp (45.75% vs. 48.41% vs. 49.77% no-show rate) |

## Validated Findings (Statistical Testing)

- Lead time, prior no-shows, and distance are all statistically significant drivers of no-shows (χ² tests, p < 0.001 for each).
- Reminder lift is **not uniform**: only Tier 3 (High Risk) shows a statistically significant improvement from reminders (p = 0.020, +6.85 percentage points). Tiers 1, 2, and 4 show no meaningful effect.
- Findings are stable across gender and appointment type — the lead-time effect isn't an artifact of one subgroup.

## Business Insights

1. Long lead times are the biggest lever — patients booked more than 30 days out are more than twice as likely to no-show as those booked within a week.
2. Reminder messaging, as currently deployed, wastes roughly 60% of its budget on segments where it has no measurable effect.
3. Distance becomes a real barrier only past 20km, suggesting a targeted (not universal) transport or telehealth intervention.

## Recommendations

1. **Target reminders to Tier 2–3 patients only** — redirect the ~60% of wasted spend toward higher-touch outreach for high-risk patients.
2. **Cap long-horizon bookings or actively re-confirm them** — apply the 14-day booking cap policy; even 50% compliance recovers ~391 slots (-7.82pp no-shows).
3. **Offer telehealth or transport support for patients beyond 20km** — addresses the distance-driven no-show spike directly.
4. **Combine both policies** — Policy A (booking cap) + Policy B (selective overbooking of Tiers 3–4) together reclaim 1,048 slots, dropping no-shows from 48.46% to 27.50%.

## Analytical Limitations

- `waiting_time_minutes` was tested and pruned — no correlation with attendance (24.29 vs. 24.20 min means).
- Cancellations (5.26% of records) are administrative releases, not lost capacity, and are excluded from the no-show definition by design.
- `distance_to_clinic_km` (90 rows) and `waiting_time_minutes` (60 rows) are median-imputed; flagged with `_was_missing` indicators but introduce minor uncertainty.
- Findings are correlational, not causal — they describe *who* is likely to miss appointments, not necessarily *why*, beyond the behavioral proxies tested.
